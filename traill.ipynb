```python
from langchain_core.prompts import PromptTemplate
from langgraph.graph import StateGraph,START,END, MessagesState
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langchain_core.messages import HumanMessage
from langgraph.checkpoint.memory import InMemorySaver
load_dotenv()
llm = ChatOpenAI(model = "gpt-4o-mini")
```

```python
class gg(MessagesState):
    a: str
```

```python
def node1(state: gg):
    result = llm.invoke("hi my name is keerthi")
    return {'messages' :[result]}
```

```python
def node2(state: gg):
    messages = state['messages']+ [HumanMessage(content = "what is my name")]
    result = llm.invoke(messages)
    print(result)
    return {'messages' : [result]}
```

```python
builder = StateGraph(gg)
builder.add_node('node1', node1)
builder.add_node('node2', node2)
builder.add_edge(START, "node1")
builder.add_edge("node1", "node2")
builder.add_edge("node2", END)
```

```python
checkpointer = InMemorySaver()
```

```python
graph = builder.compile(checkpointer = checkpointer)
graph.invoke({'a':"ddd"}, config = {"configurable":{"thread_id":"12"}})
```

```python
from langchain_core.messages import HumanMessage
from langgraph.graph import START, END, MessagesState, StateGraph
from langgraph.checkpoint.memory import InMemorySaver
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()
llm= ChatOpenAI(model = "gpt-4o-mini")
class gg(MessagesState):
    g: str
def node1(state:gg):
    message = state['messages']+[state['g']]
    r=llm.invoke(message)
    return {'messages' :[r]}
builder = StateGraph(gg)
builder.add_node("node1", node1)
builder.add_edge(START, "node1")
builder.add_edge("node1", END)
checkpointer = InMemorySaver()
graph = builder.compile(checkpointer = checkpointer)
graph.invoke({"g": "hi i am keerthi", "messaages" :[]}, config = {'configurable':{'thread_id':'56'}})
```

```python
from langgraph.graph import StateGraph, START, END, MessagesState
from langchain_core.messages import HumanMessage
from dotenv import load_dotenv
load_dotenv()
from langchain_openai import ChatOpenAI
from langgraph.checkpoint.memory import InMemorySaver
llm = ChatOpenAI(model = "gpt-4o-mini")
class hh(MessagesState):
    pass
def node1(state:hh):
    r =llm.invoke(state['messages'])
    return {'messages': [r]}
builder = StateGraph(hh)
builder.add_node("hh", node1)
builder.add_edge(START,"hh")
builder.add_edge("hh", END)
checkpointer = InMemorySaver()
graph = builder.compile(checkpointer = checkpointer)
graph.invoke({'messages':[HumanMessage("I am keerthi")]}, config = {'configurable':{'thread_id':'88'}})
graph.invoke({'messages':[HumanMessage('what is my name')]}, config = {'configurable':{'thread_id':'88'}})
```
